```python
import requests
import pandas as pd
#import paho.mqtt.client as mqtt


def get_json(endpoint):
	try:
		response = requests.get(endpoint, timeout=30) # Send a web request to the schedule URL and stop waiting after 30 seconds.
		response.raise_for_status()                   # Raise an error if the server returned an unsuccessful status code. this would look like 404 or 500. If the server returned a successful status code, this line does nothing.
		return response.json() 		                  # Read the response body as JSON data and return it to the caller.
	except requests.RequestException as exc:          # Catch request-related errors so they do not crash the whole program.
		print(f"Request failed: {exc}")               # Show the error message to help explain why the download failed.
		return None

Raw_data_path = "../../EMF_Camp/Raw_data/"
```

# Scedule data

```python
Schedule_URL = "https://www.emfcamp.org/schedule/2026.json"
# MQTT_Broker = "mqtt://mqtt.emf.camp" # + emf/weather/ + humidity etc    # API now defunct. using database dump

json_data = get_json(Schedule_URL)
if json_data is not None:
	df = pd.DataFrame(json_data)
	print(df.info())
df.to_csv(Raw_data_path + 'schedule.csv', index=False)
```

```python
import json
import requests

BASE = "https://stages.emf.camp"

# 1) List sessions
sessions = requests.get(f"{BASE}/api/sessions?limit=100", timeout=30).json()
print(json.dumps(sessions, indent=2)[:2000])

# 2) Fetch transcript for one venue
venue = "stage-a"
hist = requests.get(
    f"{BASE}/api/captions/history/{venue}",
    params={"limit": 1000, "order": "asc"},
    timeout=30
).json()

with open("camptions_history.json", "w") as f:
    json.dump(hist, f, indent=2)

print("Saved JSON to camptions_history.json")
```
